```python
from optimum.onnxruntime import ORTModelForSeq2SeqLM
from transformers import AutoTokenizer
from pathlib import Path
```

```python
def export_model(model_name, output_path):
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = ORTModelForSeq2SeqLM.from_pretrained(model_name, export=True)

    output_dir = Path(output_path)
    output_dir.mkdir(parents=True, exist_ok=True)

    model.save_pretrained(output_dir)
    tokenizer.save_pretrained(output_dir)
```

```python
# distilled 모델 선택
model_id = "facebook/nllb-200-distilled-600M"

print("🔄 Exporting distilled NLLB model...")
export_model(model_id, "assets/nllb-distilled")
print("✅ Saved: assets/nllb-distilled")
```

```python
import os

def get_file_size(path):
    size_bytes = os.path.getsize(path)
    size_gb = size_bytes / (1024 ** 3)
    return round(size_gb, 2)

def list_model_sizes(folder):
    for filename in os.listdir(folder):
        if filename.endswith(".bin") or filename.endswith(".onnx") or filename.endswith(".safetensors"):
            full_path = os.path.join(folder, filename)
            size = get_file_size(full_path)
            print(f"{filename}: {size} GB")

# 예시: 경량화 후 모델 폴더
list_model_sizes("assets/nllb-distilled")
```
