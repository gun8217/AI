```python
import os
import gdown
```

```python
# 구글 드라이브 폴더 ID
folder_id = "1etADjW0Lq_0Ow4m4AkmfrR3TMlwO4W-2"

# 다운로드할 로컬 경로
local_dir = "./"
os.makedirs(local_dir, exist_ok=True)
```

```python
assets_dir = "./assets"

files = []
for root, dirs, filenames in os.walk(assets_dir):
    for f in filenames:
        files.append(os.path.join(root, f))

print(files)
```

```python
for fname in files:
    filename = os.path.join(local_dir, fname)
    if os.path.exists(filename):
        print(f"✅ 이미 있음: {filename} → 건너뜀")
    else:
        print(f"⬇️ 새 파일 필요: {filename}")
```

```python
# gdown으로 폴더 내 파일 다운로드
files = gdown.download_folder(id=folder_id, output=local_dir, quiet=True, use_cookies=False)

# 파일 존재 여부 + 용량 비교
for f in files:
    filename = os.path.join(local_dir, os.path.basename(f))
    if os.path.exists(filename):
        local_size = os.path.getsize(filename)
        remote_size = os.path.getsize(f)  # gdown이 받은 파일 경로 기준
        if local_size != remote_size:
            print(f"⚠️ 파일 크기 다름: {filename} (local={local_size}, remote={remote_size}) → 덮어쓰기")
            # 다시 다운로드하려면 파일 ID 기반으로 gdown.download() 호출 필요
        else:
            print(f"✅ 동일 파일: {filename} (size={local_size}) → 건너뜀")
    else:
        print(f"⬇️ 새 파일 다운로드: {filename}")
```
