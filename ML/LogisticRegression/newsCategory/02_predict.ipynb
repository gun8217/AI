```python
import joblib
```

```python
# 학습된 모델과 벡터라이저 불러오기
vectorizer = joblib.load("tfidf_vectorizer.pkl")
model = joblib.load("text_classifier.pkl")
```

```python
# 분류할 실제 기사 텍스트
article = """
Jennifer Lopez Shares Her Morning Routine And Skincare Secrets

In a recent interview, Jennifer Lopez opened up about her daily beauty rituals, including her favorite moisturizers, facial massages, and how she maintains glowing skin despite a busy schedule. She also emphasized the importance of hydration and sleep in her wellness journey.
"""
```

```python
# 텍스트 벡터화
X_input = vectorizer.transform([article])
```

```python
# 예측 수행
predicted_label = model.predict(X_input)[0]
predicted_proba = model.predict_proba(X_input)[0]
```

```python
# 결과 출력
print(f"예측된 카테고리: {predicted_label}")
print("카테고리별 확률:")
for label, prob in zip(model.classes_, predicted_proba):
    print(f"  {label}: {prob:.4f}")
```
